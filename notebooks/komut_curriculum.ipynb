## 9. Manevra curriculum'u — süre hedefli Δ komutları (0–100 kt)

Aynı arayüz (Δheading / Δhız / Δirtifa) ama:

* her komutun bir **süre hedefi** T'si var: `T = tepki + max(|Δψ| / dönüş hızı, |Δv| / ivme, |Δh| / tırmanış)`,
  dönüş hızı = min(pedal hızı, g·tan(yatış)/V); **son sınır** = max(1.25·T, T+1 s),
* ajan **attitude komutu** verir: yatış (trim ± 60°) ve yunuslama (trim ∓ 30°) komutunu 133 Hz'lik bir iç döngü
  (ACAH) cyclic'e çevirir; collective ± 0.25, pedal ± 0.7; JSBSim AFCS yalnızca sönümleme,
* trim hıza göre JSBSim `steady_flight_data.xml` tablolarından; başlangıç 0–100 kt, 600–1500 ft,
* başarı: üç eksen birlikte bantta (2°, 2 ft/s, 10 ft) son sınırdan önce + 5 s; kuplaj 8°, 8 ft/s, 40 ft.

Seviyeler: **M1** (0–30 kt, rahat) → **M2** (düşük hız çeviklik: pedal dönüşü, bob-up) → **M3** (40–80 kt, 30° yatış)
→ **M4** (tüm zarf, 45° yatış) → **M5** (agresif + birleşik, 55° yatış). Claude'un koşusu: 1.68 M adım, 33 dk (2 çekirdek).
Ayrıntılar: README bölüm 28.

## 10. Dayanıklılık — zorlayıcı episode'lar ve test senaryoları

hızlanırken ani fren, art arda hızlı komutlar) ve **zarf sınırları** (hover, 100 kt, 250 ft tabanı).

* Yeni Δ o anki **ölçülen** duruma göre uygulanır; verilmeyen eksen önceki hedefine devam eder.
* Kesilen komut süre hedefiyle yargılanmaz (yalnızca kuplaj); kesen komutun süre hedefine ters yöndeki hareketi

  bu yüzden ~45° yatışta kalıyordu. Yetki ±0.45'e genişletildi: `widen_collective.py` ağırlıkları yeniden ölçekler
  (eski yetki içinde davranış birebir aynı), model yeni ayarı zip'inde taşır; sonra S4'te ince ayar.

## 11. Kalkış, hover ve iniş — dört kumanda doğrudan

Rotor warm-up'tan sonra yerden kalkış (12–1000 ft), hover, hover manevraları (yerinde dönüş, ileri / geri / yana kayma,
bob-up / bob-down), kalkış sürerken hedef değişikliği, pad'e yumuşak iniş, ağırlık / CG (8500–10280 lbs) ve hover'da
bozucular (kumanda darbesi, itki, attitude sapması).

* Ajan **dört kumandanın dördünü** doğrudan kullanır: collective, longitudinal cyclic, lateral cyclic, pedal (hover trimi
  etrafında tam yetki). AFCS yalnızca SAS (rate damping); attitude / heading hold yok, hiçbir kumanda sınırlanmadı.
* Neden dördü de gerekli: cyclic / pedal sabitken collective'i artırınca helikopter kalkışta dönüp pad'den kayıyor
  (tork → yaw, tail rotor itkisi → yana kayma). Ajan kalkışta pedal ile collective'i birlikte kullanıyor (tork telafisi).
* Başarı: son sınırdan önce hover bandına girip tutmak (konum ±6–12 ft, irtifa ±3–10 ft, heading ±3°, hız ≤ 2 ft/s);
  iniş: dört kızak noktası yerde, ağırlığın ≥ %70'i kızaklarda, pad'e ≤ 8 ft, temas ≤ 4 ft/s.
* Seviyeler K1 → K9 (K7a: iniş okulu; yerde hafif yüklü / çok alçak hover başlangıçları), eski seviyelerden tekrar
  (rehearsal). Ayrıntılar, eğitim süreci ve sonuçlar: README bölüm 30.
